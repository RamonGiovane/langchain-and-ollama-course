## Loading a PDF

````python
from langchain_community.document_loaders import PyMuPDFLoader


loader = PyMuPDFLoader('rag-dataset/health supplements/1. dietary supplements - for whom.pdf')

docs = loader.load()

print('Total pages:', len(docs))
````

````python
## Print the metadata of the first page
print(docs[0].metadata)

## Print the first page of the document
print(docs[0].page_content)
````

## Loading multiple PDFs with `os.walk`

We are going to read all the PDF documents in the sub directories of `rag-dataset` folder, recursively.

````python
import os

pdfs = []
for root, dirs, files in os.walk('rag-dataset'):
    for file in files:
        if file.endswith('.pdf'):
            pdfs.append(os.path.join(root, file))

print(f'Files (path) found: {(pdfs)}')

print('Loading documents...')

docs = []
for pdf in pdfs:
    loader = PyMuPDFLoader(pdf)
    docs.extend(loader.load())

print('Total pages in all PDFs:', len(docs))
````
